# 批量查询

```python
data = pd.read_csv("/data/rui/Packages/astrokit/demo/data/demo_data_legacysurvey.csv")
res = ls10.find_bricks_from_list(
    ra=data["ra"],
    dec=data["dec"],
    search_radius=5*60,
    max_workers=32,
    check_file_exists=True,
    show_progress=True
)
res
```

```python
res = ls10.collect_matches(
    ra=data["ra"],
    dec=data["dec"],
    search_radius=5*60,
    id=None,
    max_workers=32,
    show_progress=True
)
```
